```python
import pandas as pd
import numpy as np
import datetime as dt

# Direct UCI Repository URL se dataset load karein (CSV/Excel format)
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/00352/Online%20Retail.xlsx'
df = pd.read_excel(url)

print("Dataset Loaded Successfully!")
print(f"Total Rows & Columns: {df.shape}")
```

```python
# 1. Missing CustomerIDs waale rows filter karo
df = df.dropna(subset=['CustomerID'])

# 2. Cancelled/Returned orders (Negative Quantity/UnitPrice) remove karo
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]

# 3. Total Transaction Value (Sales) column add karo
df['TotalSum'] = df['Quantity'] * df['UnitPrice']

# Date column ko Proper DateTime format me convert karo
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

print(f"Cleaned Rows: {df.shape[0]}")
print(f"Unique Customers Count: {df['CustomerID'].nunique()}")
```

```python
# Dataset ke last transaction date se 1 din aage ki reference snapshot date
snapshot_date = df['InvoiceDate'].max() + dt.timedelta(days=1)

# GroupBy karke R, F, M Engine Build karo
rfm = df.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days, # Recency (Kitne din pehle buy kiya)
    'InvoiceNo': 'nunique',                                  # Frequency (Total Kitni baar buy kiya)
    'TotalSum': 'sum'                                        # Monetary (Total Kitna spend kiya)
}).reset_index()

# Rename Columns cleanly
rfm.columns = ['CustomerID', 'Recency', 'Frequency', 'Monetary']

# Output View
rfm.head()
```

```python
# 1. Quantiles se 1-5 Score Assign karo
rfm['R_Score'] = pd.qcut(rfm['Recency'], q=5, labels=[5, 4, 3, 2, 1]) # Kam days = Higher Score (5)
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), q=5, labels=[1, 2, 3, 4, 5])
rfm['M_Score'] = pd.qcut(rfm['Monetary'], q=5, labels=[1, 2, 3, 4, 5])

# 2. Customer Segment Mapping Logic
def assign_segment(row):
    r = int(row['R_Score'])
    f = int(row['F_Score'])
    m = int(row['M_Score'])

    if (r >= 4) and (f >= 4) and (m >= 4):
        return 'Champions (Daily Regulars)'
    elif (r >= 3) and (f >= 3):
        return 'Loyal Customers (Weekly Shoppers)'
    elif (r >= 4) and (f <= 2):
        return 'New / Potential Loyalists'
    elif (r <= 2) and (f >= 3):
        return 'At Risk (Fading Regulars)'
    else:
        return 'Lost / Casual Shoppers'

# 3. Apply Function
rfm['Segment'] = rfm.apply(assign_segment, axis=1)

# 4. Check Segment Counts
print("Customer Count Per Segment:")
print(rfm['Segment'].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Styling Setup
plt.figure(figsize=(10, 5), dpi=120)
sns.set_style("whitegrid")

# Bar Plot Creation
ax = sns.countplot(
    data=rfm,
    y='Segment',
    order=rfm['Segment'].value_counts().index,
    palette='Blues_r'
)

# Title & Formatting
plt.title('Customer Segmentation Breakdown (Konbini Model)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Number of Unique Customers', fontsize=11)
plt.ylabel('Customer Behavioral Cohort', fontsize=11)

# Values on Bars
for p in ax.patches:
    width = p.get_width()
    ax.annotate(f'{int(width)}',
                (width + 20, p.get_y() + p.get_height() / 2.),
                ha='left', va='center', fontsize=10, fontweight='bold', color='#333333')

plt.xlim(0, 1600)
plt.tight_layout()
plt.show()
```

```python
# Segment Level Aggregate Performance
segment_summary = rfm.groupby('Segment').agg({
    'Recency': 'mean',
    'Frequency': 'mean',
    'Monetary': ['mean', 'sum']
}).round(2)

segment_summary
```
